```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
input_one = input('Enter a topic to know about:')

llm = ChatOpenAI(model='gpt-5-mini',api_key=os.environ.get('OPENAI_API_KEY'))
theprompt = ChatPromptTemplate.from_messages([
            ('system','You are a senior Data engineer tutor kindly explain give topic in easy manner'),
            ('human','Tell me about {thetopic}')
])
```

```python
chain = theprompt | llm

response = chain.invoke({'thetopic':input_one}).content
print(response)
```
