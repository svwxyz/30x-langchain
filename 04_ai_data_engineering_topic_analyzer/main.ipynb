```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel
from dotenv import load_dotenv
load_dotenv()
```

```python
the_input = input('Enter a topic ::')
print(the_input)
```

```python
prompt_one = ChatPromptTemplate.from_messages([

        ('system','You are data engineering topic analyzer simply analyze the topics'),
        ('human','Explain {topic} in beginner-friendly language.')
])

llm = ChatOpenAI(model='gpt-5-mini')

chain_one = prompt_one | llm 
```

```python
prompt_two = ChatPromptTemplate.from_messages([
        ('system','Act like interviewer and ask me one question'),
        ('human','Give one beginner-level interview question about {topic}.')
])

chain_two = prompt_two | llm
```

```python
prompt_three = ChatPromptTemplate.from_messages([

        ('system','As a senior data engineer give 2-3 examples '),
        ('human','Give two real-world Data Engineering use cases of {topic}.')
])
chain_three = prompt_three | llm
```

```python
parra_chain = RunnableParallel(one=chain_one,two=chain_two,three=chain_three)

response = parra_chain.invoke({'topic':the_input})
```

```python
print(response['one'].content)
```
