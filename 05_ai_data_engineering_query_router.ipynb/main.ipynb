```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableBranch
from dotenv import load_dotenv
load_dotenv()
```

```python
the_topic = input('Enter a topic you want to ask from:')
the_question = input('Enter your question')
```

```python
print(f"Topic: {the_topic}")
print(f"Question: {the_question}")
```

```python
spark_chain = ChatPromptTemplate.from_messages([
        ('system','You are a data engineer who is an expert in Spark. You will answer questions related to Spark.'),
        ('human','I want to ask you a question about Spark. Here is the question: {question}'),
])

sql_chain = ChatPromptTemplate.from_messages([
        ('system','You are a data engineer who is an expert in SQL. You will answer questions related to SQL.'),
        ('human','I want to ask you a question about SQL. Here is the question: {question}'),
])

python_chain = ChatPromptTemplate.from_messages([
        ('system','You are a data engineer who is an expert in Python. You will answer questions related to Python.'),
        ('human','I want to ask you a question about Python. Here is the question: {question}'),
])
```

```python
llm = ChatOpenAI(model='gpt-5-mini')

chain1 = spark_chain | llm
chain2 = sql_chain | llm
chain3 = python_chain | llm
```

```python
the_router = RunnableBranch(
    (lambda x : x['topic'].lower() == 'spark', chain1),
    (lambda x : x['topic'].lower() == 'sql', chain2),
    (lambda x : x['topic'].lower() == 'python', chain3),
    llm
)
```

```python
response = the_router.invoke({'topic': the_topic, 'question': the_question})
```

```python
print(response.content)
```
