```python
from langchain_openai import ChatOpenAI
from pydantic import BaseModel , Field
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os 

load_dotenv()
```

```python
input_one = input('Enter a topic to be asked question from:')
print(input_one)
```

```python
question_prompt = ChatPromptTemplate.from_messages([

        ('system','You are a interviewer so ask one simple question about the  {thetopic}'),
])

llm = ChatOpenAI(model='gpt-5-mini',api_key=os.environ.get('OPENAI_API_KEY'))

chain_one = question_prompt | llm 

question = chain_one.invoke({'thetopic':input_one}).content
print(question)
```

```python
input_two = input('Submit your answer:')
print(input_two)
```

```python
class struct(BaseModel):
    correct:str = Field(description='Tell what terms u liked it in the ans which was correct')
    missed:str = Field(description='Tell me which terms are missed in the answer which could make it better ')
    score:str = Field(description='Give me score for my answer out of 100')
```

```python
review_prompt = ChatPromptTemplate.from_messages([
        ('system','As per this question {question} review it and based on this give '),
        ('human','{theanswer}'),
])

struc_llm = llm.with_structured_output(struct)

chain_two = review_prompt | struc_llm 

result = chain_two.invoke({'question':question,'theanswer':input_two})
```

```python
print(result.correct)
```

```python
print(result.missed)
```

```python
print(result.score)
```
