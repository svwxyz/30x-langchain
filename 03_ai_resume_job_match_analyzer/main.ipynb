```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel , Field 
from dotenv import load_dotenv 
load_dotenv()
import os
```

```python
job_description = input('Enter the Job profile desciption:')
skills = input('Enter your skills:')
```

```python
print('Job Description:', job_description)
print('Skills:', skills)
```

```python
prompts = ChatPromptTemplate.from_messages([

        ('system',
                '''Review the skills according for the job profile:
                    And then give 
                    ├── match_score
                    ├── matching_skills
                    ├── missing_skills
                    ├── strengths
                    ├── weaknesses
                    └── recommendation
                '''),
        ('human','''
        
                Here is the job description for the job : 
                    {thejob}
                and 
                Here is the skills of individual:
                    {theskilss}        
        ''')
])

llm = ChatOpenAI(model='gpt-5-mini',api_key=os.environ.get('OPENAI_API_KEY'))
```

```python
class review(BaseModel):

    score : int = Field(description='Give score on the basis of skills and job description out of 100')
    matched_skills: str = Field(description='What skills were matched')
    missed_skills: str = Field(description='Mention skills which are missing')
    strengths: str = Field(description='Mention the strengths of the individual')
    weaknesses: str = Field(description='Mention the weaknesses of the individual')
    recommendation: str = Field(description='Give recommendation to the individual')
```

```python
struc_llm = llm.with_structured_output(review)


chain = prompts | struc_llm
```

```python
response = chain.invoke({'thejob':job_description,'theskilss':skills}) 
```
